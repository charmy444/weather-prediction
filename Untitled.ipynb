## ACF/PACF: автокорреляция температуры

ACF показывает общую зависимость с прошлыми значениями, а PACF помогает отделить прямой вклад конкретного лага. Чтобы графики были читаемыми, агрегируем почасовые данные до дневной средней температуры.

```python
daily_temp = (
    final_df
    .set_index("ds")
    .groupby("city")["temperature_2m"]
    .resample("D")
    .mean()
    .reset_index()
)

cities = daily_temp["city"].unique()
fig, axes = plt.subplots(len(cities), 2, figsize=(16, 3 * len(cities)))

for i, city in enumerate(cities):
    series = daily_temp.loc[daily_temp["city"] == city, "temperature_2m"].dropna()
    plot_acf(series, ax=axes[i, 0], lags=60)
    plot_pacf(series, ax=axes[i, 1], lags=60, method="ywm")
    axes[i, 0].set_title(f"{city}: ACF")
    axes[i, 1].set_title(f"{city}: PACF")

plt.tight_layout()
plt.show()
```

## Feature Engineering для прогноза на 7 дней

Добавляем краткосрочные лаги, недельную сезонность, годовую сезонность, скользящие статистики, календарные и Fourier-признаки. Такой набор подходит и для классических моделей, и для бустинга/случайного леса.

```python
final_df = final_df.sort_values(["city", "ds"]).copy()

# Лаги: краткосрочная динамика, недельная и годовая сезонность.
lags = {
    "1h": 1,
    "2h": 2,
    "6h": 6,
    "12h": 12,
    "24h": 24,
    "48h": 48,
    "72h": 72,
    "7d": 24 * 7,
    "30d": 24 * 30,
    "365d": 24 * 365,
}

for name, hours in lags.items():
    final_df[f"temperature_2m_lag_{name}"] = (
        final_df.groupby("city")["temperature_2m"].shift(hours)
    )

for window in [24, 24 * 7, 24 * 30]:
    shifted = final_df.groupby("city")["temperature_2m"].shift(1)
    final_df[f"temperature_2m_roll_mean_{window}h"] = (
        shifted.groupby(final_df["city"]).transform(lambda s: s.rolling(window, min_periods=24).mean())
    )
    final_df[f"temperature_2m_roll_std_{window}h"] = (
        shifted.groupby(final_df["city"]).transform(lambda s: s.rolling(window, min_periods=24).std())
    )

# Календарные признаки.
final_df["hour"] = final_df["ds"].dt.hour
final_df["dayofweek"] = final_df["ds"].dt.dayofweek
final_df["month"] = final_df["ds"].dt.month
final_df["dayofyear"] = final_df["ds"].dt.dayofyear

# Циклическое кодирование календаря.
final_df["hour_sin"] = np.sin(2 * np.pi * final_df["hour"] / 24)
final_df["hour_cos"] = np.cos(2 * np.pi * final_df["hour"] / 24)
final_df["doy_sin"] = np.sin(2 * np.pi * final_df["dayofyear"] / 365.25)
final_df["doy_cos"] = np.cos(2 * np.pi * final_df["dayofyear"] / 365.25)

final_df.tail()
```

## Стационарность: ADF и KPSS

Проверяем дневные ряды температуры по городам. ADF проверяет наличие единичного корня, KPSS - стационарность относительно уровня/тренда. Вместе они дают более надежную картину, чем один тест.

```python
def stationarity_tests(series):
    series = series.dropna()
    adf_stat, adf_pvalue, *_ = adfuller(series, autolag="AIC")
    kpss_stat, kpss_pvalue, *_ = kpss(series, regression="c", nlags="auto")
    return pd.Series({
        "adf_pvalue": adf_pvalue,
        "kpss_pvalue": kpss_pvalue,
    })

stationarity_report = (
    daily_temp
    .groupby("city")["temperature_2m"]
    .apply(stationarity_tests)
    .unstack()
)
stationarity_report
```

## STL-декомпозиция: тренд, сезонность, остатки

Для почасового ряда используем дневную агрегацию и период 365, чтобы увидеть годовую сезонность. Это дает понятную интерпретацию: базовый уровень температуры, сезонная компонента и необычные отклонения.

```python
for city in daily_temp["city"].unique():
    series = (
        daily_temp[daily_temp["city"] == city]
        .set_index("ds")["temperature_2m"]
        .asfreq("D")
        .interpolate()
    )
    result = STL(series, period=365, robust=True).fit()
    fig = result.plot()
    fig.set_size_inches(14, 7)
    fig.suptitle(f"STL-декомпозиция дневной температуры: {city}", y=1.02)
    plt.show()
```

## Дерево решений, Random Forest, CatBoost и Optuna

Сравним несколько табличных моделей на одинаковом наборе признаков. Дерево решений дает простой интерпретируемый baseline, Random Forest снижает переобучение за счет ансамбля деревьев, CatBoost хорошо работает на табличных данных, а Optuna подбирает гиперпараметры CatBoost по validation-метрике.

```python
feature_cols = [
    column for column in model_df.columns
    if column not in ["ds", "target_7d"]
]

X_train = pd.get_dummies(train_df[feature_cols], columns=["city"], drop_first=False)
X_val = pd.get_dummies(val_df[feature_cols], columns=["city"], drop_first=False)
X_test = pd.get_dummies(test_df[feature_cols], columns=["city"], drop_first=False)

# На случай если в одном из сплитов нет какого-то города/категории.
X_val = X_val.reindex(columns=X_train.columns, fill_value=0)
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)

y_train = train_df["target_7d"]
y_val = val_df["target_7d"]
y_test = test_df["target_7d"]

def wape(true, pred):
    true = pd.Series(true)
    pred = pd.Series(pred, index=true.index)
    return (true - pred).abs().sum() / true.abs().sum()

def regression_metrics(y_true, y_pred):
    return pd.Series({
        "MAE": mean_absolute_error(y_true, y_pred),
        "RMSE": mean_squared_error(y_true, y_pred) ** 0.5,
        "WAPE": wape(y_true, y_pred),
    })
```

```python
tree_model = DecisionTreeRegressor(
    max_depth=12,
    min_samples_leaf=50,
    random_state=42,
)
tree_model.fit(X_train, y_train)

tree_predictions = {
    "validation": tree_model.predict(X_val),
    "test": tree_model.predict(X_test),
}

tree_metrics = pd.DataFrame([
    {"model": "Decision Tree", "split": "validation", **regression_metrics(y_val, tree_predictions["validation"]).to_dict()},
    {"model": "Decision Tree", "split": "test", **regression_metrics(y_test, tree_predictions["test"]).to_dict()},
])
tree_metrics
```

```python
forest_model = RandomForestRegressor(
    n_estimators=80,
    max_depth=16,
    min_samples_leaf=20,
    n_jobs=-1,
    random_state=42,
)
forest_model.fit(X_train, y_train)

forest_predictions = {
    "validation": forest_model.predict(X_val),
    "test": forest_model.predict(X_test),
}

forest_metrics = pd.DataFrame([
    {"model": "Random Forest", "split": "validation", **regression_metrics(y_val, forest_predictions["validation"]).to_dict()},
    {"model": "Random Forest", "split": "test", **regression_metrics(y_test, forest_predictions["test"]).to_dict()},
])
forest_metrics
```

```python
catboost_model = CatBoostRegressor(
    iterations=500,
    learning_rate=0.05,
    depth=8,
    loss_function="RMSE",
    random_seed=42,
    verbose=False,
    allow_writing_files=False,
)
catboost_model.fit(X_train, y_train)

catboost_predictions = {
    "validation": catboost_model.predict(X_val),
    "test": catboost_model.predict(X_test),
}

catboost_metrics = pd.DataFrame([
    {"model": "CatBoost", "split": "validation", **regression_metrics(y_val, catboost_predictions["validation"]).to_dict()},
    {"model": "CatBoost", "split": "test", **regression_metrics(y_test, catboost_predictions["test"]).to_dict()},
])
catboost_metrics
```

## Optuna

```python
optuna.logging.set_verbosity(optuna.logging.WARNING)

tune_sample_size = min(60000, len(X_train))
tune_idx = X_train.sample(n=tune_sample_size, random_state=42).index
X_train_tune = X_train.loc[tune_idx]
y_train_tune = y_train.loc[tune_idx]

def objective(trial):
    params = {
        "iterations": trial.suggest_int("iterations", 200, 700, step=100),
        "learning_rate": trial.suggest_float("learning_rate", 0.02, 0.15, log=True),
        "depth": trial.suggest_int("depth", 4, 10),
        "l2_leaf_reg": trial.suggest_float("l2_leaf_reg", 1.0, 20.0, log=True),
        "random_strength": trial.suggest_float("random_strength", 0.1, 5.0),
        "loss_function": "RMSE",
        "eval_metric": "MAE",
        "random_seed": 42,
        "verbose": False,
        "allow_writing_files": False,
    }

    model = CatBoostRegressor(**params)
    model.fit(
        X_train_tune,
        y_train_tune,
        eval_set=(X_val, y_val),
        use_best_model=True,
        early_stopping_rounds=50,
        verbose=False,
    )
    val_pred = model.predict(X_val)
    return mean_absolute_error(y_val, val_pred)

study = optuna.create_study(direction="minimize", study_name="catboost_temperature_forecast")
study.optimize(objective, n_trials=10, show_progress_bar=False)

study.best_value, study.best_params
```

```python
best_catboost_params = {
    **study.best_params,
    "loss_function": "RMSE",
    "eval_metric": "MAE",
    "random_seed": 42,
    "verbose": False,
    "allow_writing_files": False,
}

optuna_catboost_model = CatBoostRegressor(**best_catboost_params)
optuna_catboost_model.fit(
    X_train,
    y_train,
    eval_set=(X_val, y_val),
    use_best_model=True,
    early_stopping_rounds=50,
    verbose=False,
)

optuna_catboost_predictions = {
    "validation": optuna_catboost_model.predict(X_val),
    "test": optuna_catboost_model.predict(X_test),
}

optuna_catboost_metrics = pd.DataFrame([
    {"model": "CatBoost + Optuna", "split": "validation", **regression_metrics(y_val, optuna_catboost_predictions["validation"]).to_dict()},
    {"model": "CatBoost + Optuna", "split": "test", **regression_metrics(y_test, optuna_catboost_predictions["test"]).to_dict()},
])
optuna_catboost_metrics
```
